```python
# min/max compactness, shape_index, interior_edge, fractal_dim

def min1(col: pd.Series): 
    return col.min()

display(min1(zambia_shp["compactness"]))
display(min1(zambia_shp["shape_index"]))
display(min1(zambia_shp["interior_edge"]))
display(min1(zambia_shp["fractal_dim"]))

def max1(col: pd.Series): 
    return col.min()

display(max1(zambia_shp["compactness"]))
display(max1(zambia_shp["shape_index"]))
display(max1(zambia_shp["interior_edge"]))
display(max1(zambia_shp["fractal_dim"]))
```

```python
# range of compactness, shape_index, interior_edge, fractal_dim

def range1(col: pd.Series): 
    return col.max() - col.min()

display(range1(zambia_shp["compactness"]))
display(range1(zambia_shp["shape_index"]))
display(range1(zambia_shp["interior_edge"]))
range1(zambia_shp["fractal_dim"])

```
